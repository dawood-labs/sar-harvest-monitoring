- Copy them from QGIS (right-click on the map > Copy Coordinate > EPSG:4326) or from Google Maps.
- Either order works: `lat,lon` or `lon,lat`. The larger number is taken as the longitude (true for this area).
- For each place you get: what the original report and our results say on every report date, the SAR history (the pixel itself and the 5 x 5 mean the model sees), the optical greenness, and a figure.

```python
points = """

"""      # paste here, one place per line, as  latitude,longitude  or  longitude,latitude

area, config = "full", "configs/config_full.yaml"   # whole area; for the test tile use "test_tile", "configs/config.yaml"
```

```python
import os, re, sys
from pathlib import Path
from IPython.display import Image, display

root = Path.cwd() if (Path.cwd() / 'src').exists() else Path.cwd().parent   # works from the project folder or from notebooks/
os.chdir(root)
sys.path.insert(0, str(root / 'src'))
import inspect_pixel

for line in points.strip().splitlines():
    nums = [float(x) for x in re.findall(r'-?\d+\.\d+|-?\d+', line)]
    if len(nums) < 2:
        continue
    lat, lon = sorted(nums[:2])                 # larger value = longitude in this part of the world
    print('=' * 100)
    figure = inspect_pixel.main(str(lon), str(lat), area, config)
    if figure:
        display(Image(filename=str(figure)))
```
